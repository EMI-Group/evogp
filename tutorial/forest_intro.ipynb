```python
from evogp.tree import Tree, Forest, GenerateDescriptor
# the rules used to randomly generate the trees are described by a GenerateDescriptor
descriptor = GenerateDescriptor(
    max_tree_len=128,
    input_len=3,
    output_len=1,
    const_prob=0.5,
    out_prob=0.5,
    using_funcs={"+": 0.25, "-": 0.25, "*": 0.25, "/": 0.25},
    max_layer_cnt=5,
    layer_leaf_prob=0.2,
    const_range=(-1, 1),
    sample_cnt=8,
)
# random generate a Forest with 10 trees
forest = Forest.random_generate(pop_size=10, descriptor=descriptor)
```

```python
# the forest can be print or transform to a str
print(forest)
string_forest = str(forest)
print(string_forest)
```

```python
# the forest support __getitem__ method

# get one item from the forest, will obtain a tree
tree = forest[0]
print(tree)
```

```python
# get multiple items from forest, will obtain another forest

sub_forest = forest[:3]
print(sub_forest)
```

```python
# the forest support __setitem__ method

# set a Tree in the forest
forest[-1] = tree
print(forest)
```

```python
# set a forest in the forest
forest[-3:] = sub_forest
forest
```

```python
# forest support __add__ operation
forest = forest + tree
forest = forest + sub_forest
forest
```
